```python
X = df.copy()
y = df['status_type'].copy()
```

```python
from sklearn.preprocessing import LabelEncoder

le = LabelEncoder()

X['status_type'] = le.fit_transform(X['status_type'])

y = le.transform(y)
```

```python
cols = X.columns
```

```python
from sklearn.preprocessing import MinMaxScaler

ms = MinMaxScaler()

X = ms.fit_transform(X)
```

```python
X = pd.DataFrame(X, columns=cols)
```

```python
X.head()
```

```python
from sklearn.cluster import KMeans

kmeans = KMeans(n_clusters=2, random_state=0)

kmeans.fit(X)
```

```python
clusters = kmeans.predict(X)
```

```python
kmeans.cluster_centers_
```

```python
kmeans.inertia_
```

```python
labels = kmeans.labels_

# check how many of the samples were correctly labeled
correct_labels = sum(y == labels)

print("Result: %d out of %d samples were correctly labeled." % (correct_labels, y.size))
```

```python
print('Accuracy score: {0:0.2f}'. format(correct_labels/float(y.size)))
```

```python
from sklearn.cluster import KMeans
cs = []
for i in range(1, 11):
    kmeans = KMeans(n_clusters = i, init = 'k-means++', max_iter = 300, n_init = 10, random_state = 0)
    kmeans.fit(X)
    cs.append(kmeans.inertia_)
plt.plot(range(1, 11), cs)
plt.title('The Elbow Method')
plt.xlabel('Number of clusters')
plt.ylabel('CS')
plt.show()
```

```python
from sklearn.metrics import silhouette_score

silhouette_avg = silhouette_score(X, clusters)
print(f"The average silhouette score is: {silhouette_avg:.2f}")
```

```python
from sklearn.cluster import DBSCAN
from sklearn.metrics import silhouette_score

dbscan = DBSCAN(eps=0.5, min_samples=5)  # Default values, can be tuned
dbscan_clusters = dbscan.fit_predict(X)

# Calculate silhouette score for DBSCAN
# Silhouette score requires at least 2 clusters and not all points belonging to noise (-1)
if len(set(dbscan_clusters)) > 1 and -1 not in dbscan_clusters:
    silhouette_avg_dbscan = silhouette_score(X, dbscan_clusters)
    print(f"The average silhouette score for DBSCAN is: {silhouette_avg_dbscan:.2f}")
else:
    print("Cannot calculate silhouette score for DBSCAN: insufficient number of clusters or all points are noise.")

# You can also inspect the number of clusters found by DBSCAN (excluding noise points)
num_dbscan_clusters = len(set(dbscan_clusters)) - (1 if -1 in dbscan_clusters else 0)
print(f"Number of clusters found by DBSCAN: {num_dbscan_clusters}")
print(f"Number of noise points (-1 label): {list(dbscan_clusters).count(-1)}")
```

```python
from sklearn.neighbors import NearestNeighbors
import matplotlib.pyplot as plt
import numpy as np

# Calculate k for min_samples (e.g., 2 * number of features)
k = 2 * X.shape[1] # X.shape[1] is the number of features
print(f"Using k = {k} for NearestNeighbors, which will be our min_samples value.")

# Find the k-th nearest neighbor distance for each point
neigh = NearestNeighbors(n_neighbors=k)
neigh.fit(X)
distances, indices = neigh.kneighbors(X)

# Sort the distances in ascending order
distances = np.sort(distances[:, k-1], axis=0)

# Plot the k-distance graph
plt.figure(figsize=(12, 6))
plt.plot(distances)
plt.title(f'K-distance Graph (k={k})')
plt.xlabel('Data points sorted by distance')
plt.ylabel(f'{k}-th Nearest Neighbor Distance')
plt.grid(True)
plt.show()

print("Look for the 'elbow' in the graph to determine an appropriate `eps` value.")
```

```python
from sklearn.cluster import DBSCAN
from sklearn.metrics import silhouette_score

# Set min_samples based on the heuristic (2 * number of features)
min_samples_val = 2 * X.shape[1] # X.shape[1] is the number of features
eps_val = 0.28 # Selected from the k-distance graph

dbscan = DBSCAN(eps=eps_val, min_samples=min_samples_val)
dbscan_clusters_tuned = dbscan.fit_predict(X)

# Filter out noise points for silhouette score calculation
clustered_indices = dbscan_clusters_tuned != -1
X_clustered = X[clustered_indices]
dbscan_labels_clustered = dbscan_clusters_tuned[clustered_indices]

# Calculate silhouette score for DBSCAN
# Silhouette score requires at least 2 clusters among the non-noise points
if len(set(dbscan_labels_clustered)) > 1:
    silhouette_avg_dbscan_tuned = silhouette_score(X_clustered, dbscan_labels_clustered)
    print(f"The average silhouette score for DBSCAN (eps={eps_val}, min_samples={min_samples_val}) on clustered points is: {silhouette_avg_dbscan_tuned:.2f}")
else:
    print(f"Cannot calculate silhouette score for DBSCAN (eps={eps_val}, min_samples={min_samples_val}): insufficient number of clusters (or all points are noise) among non-noise points.")

# Inspect the number of clusters found by DBSCAN (excluding noise points)
num_dbscan_clusters_tuned = len(set(dbscan_clusters_tuned)) - (1 if -1 in dbscan_clusters_tuned else 0)
print(f"Number of clusters found by DBSCAN: {num_dbscan_clusters_tuned}")
print(f"Number of noise points (-1 label): {list(dbscan_clusters_tuned).count(-1)}")
```
